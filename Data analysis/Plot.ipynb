```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

beginner = [5, 4, 8, 3, 5, 12, 11, 6]
intermediate = [15, 13, 16, 6, 13, 12, 10]
expert = [16, 16, 21, 17, 16, 20]


data = [beginner, intermediate, expert]

X = np.array([0]*len(beginner) + [1]*len(intermediate)+ [2]*len(expert))
y = np.array(beginner + intermediate + expert)
res = stats.spearmanr(X,y)
print(f"coef: {res.statistic} p-value: {res.pvalue}")


plt.boxplot(data)


plt.xticks([1, 2, 3], ["Beginner", "Intermediate", "Expert"])
plt.ylabel("Number of Chess Puzzles Solved")
plt.title("Number of Chess Puzzles Solved by Skill Level")


plt.grid(True)
plt.show()
```
